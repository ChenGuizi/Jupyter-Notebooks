```python
import os
from dotenv import load_dotenv
from google import genai
# Load variables from your environment file
load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")
if not api_key:
    raise ValueError("Missing GEMINI_API_KEY in your environment file")
```

```python
client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model="gemini-3.5-flash-lite", #"gemini-3.5-flash"
    contents="Write a short greeting and explain what Gemini is."
)

print(response.text)
```
